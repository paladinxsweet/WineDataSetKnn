```python
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.
from sklearn.metrics import classification_report,accuracy_score,confusion_matrix
from sklearn import metrics
from sklearn import preprocessing
from matplotlib.backends.backend_pdf import PdfPages
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
wine=load_wine()
```

```python
X=wine.data
y=wine.target
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
Ks=10
mean_acc=np.zeros((Ks-1))
std_acc=np.zeros((Ks-1))

for n in range(1,Ks):
    neigh=KNeighborsClassifier(n_neighbors=n).fit(X_train,y_train)
    yhat=neigh.predict(X_test)
    mean_acc[n-1]=metrics.accuracy_score(y_test,yhat)

    std_acc[n-1]=np.std(yhat==y_test)/np.sqrt(yhat.shape[0])

mean_acc
```

```python
print("The best accuracy was with",mean_acc.max(),"with k=",mean_acc.argmax()+1)
```

```python
k=5
neigh5=KNeighborsClassifier(n_neighbors=k).fit(X_train,y_train)
yhat5=neigh5.predict(X_test)
```

```python
train_accuracy=accuracy_score(y_train,neigh5.predict(X_train))
test_accuracy=accuracy_score(y_test,yhat5)

print("train accuracy",train_accuracy)
print("test_accuracy",test_accuracy)
```

```python
report=classification_report(y_test,yhat5,target_names=wine.target_names)
```

```python
cm=confusion_matrix(y_test,yhat5)
```

```python
with PdfPages('knn_wine_results.pdf') as pdf:
    
    # Sayfa 1: Model Doğruluk Oranları
    plt.figure(figsize=(8.5, 11))
    plt.axis('off')  # Ekseni kaldır
    plt.text(0.1, 0.9, f"KNN Model Sonuçları (k={k}) - Wine Veri Seti", fontsize=14, weight='bold', va='top')
    plt.text(0.1, 0.8, f"Train Set Accuracy: {train_accuracy:.2f}\nTest Set Accuracy: {test_accuracy:.2f}",
             fontsize=12, va='top', family='monospace')
    pdf.savefig()
    plt.close()
    
    # Sayfa 2: Classification Report
    plt.figure(figsize=(8.5, 11))
    plt.axis('off')
    plt.text(0.1, 0.9, "Classification Report:", fontsize=14, weight='bold', va='top')
    plt.text(0.1, 0.8, report, fontsize=10, va='top', family='monospace')
    pdf.savefig()
    plt.close()
    
    # Sayfa 3: Karmaşıklık Matrisi Isı Haritası
    plt.figure(figsize=(8, 6))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=wine.target_names, 
                yticklabels=wine.target_names)
    plt.xlabel('Tahmin Edilen Sınıf')
    plt.ylabel('Gerçek Sınıf')
    plt.title('Karmaşıklık Matrisi')
    pdf.savefig()  # PDF'ye ekle
    plt.close()
    
print("Sonuçlar PDF dosyasına kaydedildi: 'knn_wine_results.pdf'")
```
